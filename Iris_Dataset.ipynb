```python
import pandas as pd

columns = [
    "SepalLength",
    "SepalWidth",
    "PetalLength",
    "PetalWidth",
    "Species"
]

df = pd.read_csv("iris/iris.data", names=columns)
```

```python
# Mean
print("Mean")
print(df.mean(numeric_only=True))

# Median
print("\nMedian")
print(df.median(numeric_only=True))

# Mode
print("\nMode")
print(df.mode(numeric_only=True))
```

```python
print("Null values in each column:")
print(df.isnull().sum())
```

```python
# Replace null values in numeric columns with their mean
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
print("\nNull values after replacement:")
print(df.isnull().sum())
```
