```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("Automobile_data.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['price'],
    bins=30,
    kde=True,
    color='skyblue'
)

plt.title("Distribution of Automobile Prices")
plt.xlabel("Price")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.boxplot(
    x='body-style',
    y='price',
    data=df
)

plt.xticks(rotation=30)
plt.title("Price by Body Style")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='horsepower',
    y='price',
    data=df
)

plt.title("Horsepower vs Price")
plt.xlabel("Horsepower")
plt.ylabel("Price")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x='fuel-type',
    hue='fuel-type',
    data=df,
    palette='Set2',
    legend=False
)

plt.title("Count of Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,6))

df['body-style'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Body Style Distribution")
plt.ylabel("")

plt.show()
```
