```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv("Titanic-Dataset.csv")

print("First Five Records")
print(df.head())

print("\nDataset Shape:", df.shape)

print("\nData Types")
print(df.dtypes)
```

```python
print("\nMissing Values")
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df["Cabin"] = df["Cabin"].fillna("Unknown")
print("\nMissing Values After Imputation")
print(df.isnull().sum())
```

```python
print("\nDuplicate Records:", df.duplicated().sum())
df = df.drop_duplicates()
print("Dataset Shape After Removing Duplicates:", df.shape)
```

```python
encoder = LabelEncoder()
df["Sex"] = encoder.fit_transform(df["Sex"])
df["Embarked"] = encoder.fit_transform(df["Embarked"])
print("\nEncoded Data")
print(df[["Sex", "Embarked"]].head())
```

```python
scaler = StandardScaler()
df[["Age", "Fare"]] = scaler.fit_transform(df[["Age", "Fare"]])
print("\nStandardized Age and Fare")
print(df[["Age", "Fare"]].head())
```

```python
normalizer = MinMaxScaler()
df[["Age", "Fare"]] = normalizer.fit_transform(df[["Age", "Fare"]])
print("\nNormalized Age and Fare")
print(df[["Age", "Fare"]].head())
```

```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1
Lower = Q1 - 1.5 * IQR
Upper = Q3 + 1.5 * IQR
outliers = df[(df["Fare"] < Lower) | (df["Fare"] > Upper)]
print("\nNumber of Outliers in Fare:", len(outliers))
```

```python
df = df[(df["Fare"] >= Lower) & (df["Fare"] <= Upper)]
print("\nDataset Shape After Removing Outliers")
print(df.shape)
```

```python
df = df.drop(["PassengerId", "Name", "Ticket", "Cabin"], axis=1)
print("\nFinal Dataset")
print(df.head())
```

```python
df.to_csv("Titanic_Preprocessed.csv", index=False)
print("\nPreprocessed dataset saved successfully.")
```
